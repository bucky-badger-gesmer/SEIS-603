```python
import pandas as pd
grades = pd.DataFrame({"Name": ["Alice", "Bob", "Charlie"], "Grade": [85, 90, 78]})
print(grades)

import matplotlib.pyplot as plt
plt.bar(grades["Name"], grades["Grade"])
plt.title("Student Grades")
plt.xlabel("Student")
plt.ylabel("Grade")
plt.show()
```
