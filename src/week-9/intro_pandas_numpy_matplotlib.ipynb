## Part 1. NumPy Basics
NumPy provides efficient array operations and mathematical functions.

```python
# Create a simple NumPy array of numbers 1 through 10
arr = np.array([(i) for i in range(1, 11)])

# Print the array and its mean
print("Array:", arr)
print("Mean:", np.average(arr))

# Create a 3x3 matrix of random numbers between 0 and 1
matrix = np.random.rand(3, 3)
print("3x3 random matrix:\n", matrix)
```

## Part 2. Pandas Basics
Pandas helps manage and analyze tabular data using DataFrames.

```python

# Create a small DataFrame with names and scores
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Eli"],
    "Score": [90, 75, 88, 92, 79]
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)

# Compute the average score
avg_score = df["Score"].mean()
print("Average Score:", avg_score)

# Add a new column called 'Passed' (True if score >= 80)
df["Passed"] = df["Score"].apply(lambda score: "Passed" if score > 80 else "Failed")
print(df)
```

## Part 3. Matplotlib Visualization
Matplotlib is used for plotting data visually.

```python

# Make a simple line plot of the scores
plt.figure(figsize=(6,4))
### YOUR CODE HERE  # (e.g., plt.plot(df["Name"], df["Score"], marker='o'))
# plt.plot(df["Name"], df["Score"], marker='o')
plt.bar(df["Name"], df["Score"], color=["green" if status == "Passed" else "red" for status in df["Passed"]])
plt.title("Student Scores")
plt.xlabel("Name")
plt.ylabel("Score")
plt.grid(True)
plt.show()

# Make a histogram of random data
data = np.random.randn(100)
# print('data', data)
plt.figure()
### YOUR CODE HERE  # (e.g., plt.hist(data, bins=10))
plt.hist(data, bins=100)
plt.title("Histogram of Random Data")
plt.show()
```
