## Pandas Series

Similar to a python list and dictionary. A series is essentially a one-dimensional labeled array, or a single column of a DataFrame.
- A pandas series associates a label with each element
- If an index is not explicitly provided by the user, pandas creates a RangeIndex ranging from 0 to N-1

Each series object also has a data type

````python
my_list = [5, 6, 7, 8]
my_series = pd.Series(my_list)

print(my_series)
````

### Labels

If nothing is specified, values are labeled by their index number.

This label can be used to access a specified value

---

### Extracting all values in series and individual elements by index

````python
my_series = pd.Series([5, 6, 7, 8, 9, 10])

print(my_series.values)
print("=" * 20)
print(my_series[4])
````

### Manual Index

````python
my_series = pd.Series(
    [5, 6, 7, 8, 9, 10],
    index=['a', 'b', 'c', 'd', 'e', 'f']
)

print(my_series)
print(my_series.values)
print("=" * 20)
print(my_series["f"])
````

### Retrieve several elements of a series by their indices or make group assignments

````python
my_series = pd.Series(
    [5, 6, 7, 8, 9, 10],
    index=['a', 'b', 'c', 'd', 'e', 'f']
)

print(my_series)
print('=' * 20)

my_series[['a', 'b', 'f']] = 0
print(my_series)
````

### Even with labels, numerical index still works

````python
my_series = pd.Series(
    [1, 7, 2],
    index=["x", "y", "z"]
)

print(my_series[[0, 1]])
````

### Creating a simple pandas series from a dictionary

````python
calories = {
    "day1": 420,
    "day2": 380,
    "day3": 390
}

my_series = pd.Series(calories)

print(my_series)
````

## Pandas Dataframe

A dataframe is simply a table
- It has rows and columns
- Sort of like a spreadhseet

Each column in a dataframe is a **series** object, with each row consisting of elements inside a series
- Tabular structure
- Labeled axis, with each row and column having labels (indices)
- Heterogenrous data
    - It can contain different data types, formats, structures or sources
 - Size-mutable, which means you can add rows and columns
 - Data alignmnet, which  means when performing operations, pandas aligns data by the row and column labels

### Making a DataFrame from a dictionary of lists

````python
data = {
    "Name": ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    "Age": [24, 27, 22, 32, 29],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]
}

# convert dictionary to a DataFrame
df = pd.DataFrame(data)

print(df)
````

**Remember:** multiple series (columns) make up a DataFrame

### Creating a DataFrame from Series

Note how in this example, both series have the same length of 3 so we want to put them side-by-side with `axis=1`

````python
series1 = pd.Series(
    [10, 20, 30],
    name='A'
)
series2 = pd.Series(
    [40, 50, 60],
    name='B'
)

df = pd.concat([series1, series2], axis=1)
print(df)
````

### First Rows and Last Rows

````python
df = pd.DataFrame(
    [23, 33, 44],
    ['Michael', 'Scottie', 'Patrick']
)

print(df)
print("=" * 20)

# By default, head and tail to the first 5 and last 5, respectively
print(df.head(1))
print(df.tail(1))
````

## Matplotlib PyPlot

````python
import matplotlib.pyplot as plt
import pandas as pd

x_test = [1, 2, 3, 4]
y_test = [1, 2, 3, 7]

df = pd.DataFrame({
    'x': x_test,
    'y': y_test
})

plt.plot(df.x, df.y)
plt.show()
````

---

````python
# Step 1: Create a Pandas DataFrame from a Python dictionary
data = {
    'A': [1, 2, 3, 4],
    'B': [5, 6, 7, 8],
    'C': [9, 10, 11, 12]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Step 2: Square all the values in the DataFrame
df_squared = df ** 2
print("\nDataFrame with squared values:")
print(df_squared)

# Step 3: Plot the squared values
df_squared.plot(kind='bar')
plt.title('Squared Values')
plt.xlabel('Index')
plt.ylabel('Value')
plt.show()
````
