## Converting Types

Handy for cleaning real-world data!

```python
data = {
    "Name": [
        "Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona", "George", "Hannah", "Ian", "Julia"
    ],
    "Age": [24, 30, 22, 28, 35, 27, 29, 31, 26, 23],
    "City": [
        "New York", "Chicago", "Seattle", "Boston", "Austin", "San Diego", "Denver", "Miami", "Houston", "Portland"
    ],
    "Score": [88, 92, 95, 78, 85, 91, 89, 87, 93, 90]
}

df = pd.DataFrame(data)

print(df)

print("type before conversion:", df["Score"].dtype)
df["Score"] = df["Score"].astype(float)
print("type after converstion:", df["Score"].dtype)

print(df)
```

## groupby() and aggregation

Notice how in this example, we group by Age and average the score! We can also do the same with City!

```python
data = {
    "Name": [
        "Alice", "Bob", "Charlie", "Diana", "Ethan",
        "Fiona", "George", "Hannah", "Ian", "Julia",
        "Kevin", "Laura", "Mike", "Nina", "Oliver",
        "Paula", "Quinn", "Rachel", "Sam", "Tina"
    ],
    "Age": [
        24, 30, 22, 28, 35,
        27, 29, 31, 26, 23,
        24, 30, 22, 28, 35,
        27, 29, 31, 26, 23
    ],
    "City": [
        "New York", "Chicago", "Seattle", "Boston", "Austin",
        "San Diego", "Denver", "Miami", "Houston", "Portland",
        "New York", "Chicago", "Seattle", "Boston", "Austin",
        "San Diego", "Denver", "Miami", "Houston", "Portland"
    ],
    "Score": [
        88, 92, 95, 78, 85,
        91, 89, 87, 93, 90,
        86, 94, 96, 80, 84,
        89, 88, 86, 94, 92
    ]
}

df = pd.DataFrame(data)

print(df)

df.groupby("Age")["Score"].mean()
```
